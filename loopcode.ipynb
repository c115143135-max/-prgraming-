# Loop
Loop指令可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。


```python
#底下的程式碼會印出0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

#  Example
計算1+2+...+10  


sum=0  
i=0 | i+1=1 | sum=sum+(i+1)-->sum = (0)+1  
i=1 | i+1=2 | sum=sum+(i+1)-->sum = (0+1)+2  
i=2 | i+1=3 | sum=sum+(i+1)-->sum = (1+2)+3  
i=3 | i+1=4 | sum=sum+(i+1)-->sum = (1+2+3)+4  

```python
sum=0
for i in range(10):
    sum=sum+(i+1)
    print(sum)
   
```

```python
for i in range(5):
    print('&'*(i+1))
```

```python
print(f'1*1={1*1}')
print(f'1*2={1*2}')
print(f'1*3={1*3}')
print(f'1*4={1*4}')  

for i in range(10):
    print(f'1*{(i+1)}={1*(i+1)}')
```
