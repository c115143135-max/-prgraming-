# Loop
for (Loop指令) 可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。


```python
#底下的程式碼會印出0-9
for i in range(10):
    print(i)
```

#  Example
計算1+2+...+10  


sum=0  
i=0 | i+1=1 | sum=sum+(i+1)-->sum = (0)+1  
i=1 | i+1=2 | sum=sum+(i+1)-->sum = (0+1)+2  
i=2 | i+1=3 | sum=sum+(i+1)-->sum = (1+2)+3  
i=3 | i+1=4 | sum=sum+(i+1)-->sum = (1+2+3)+4  

```python
sum=0
for i in range(10):
    sum=sum+(i+1)
    print(sum)
   
```

#  Range


```python
print(list(range(5)))
# 區間，由 0 開始，到 5-1 結束
print(list(range(0,5)))
# 區間，由 1 開始，到 5-1 結束
print(list(range(1,5)))
```

#  九九乘法表

  1\*1=1,1\*2=2,1\*3=3...  
  2\*1=1,2\*2=4,2\*3=6...  
  3\*1=3,3\*2=6,3\*3=9...  
  

```python
j=1
print(f'1*{j}={1*j}',end=',')
j=2
print(f'1*{j}={1*j}',end=',')
j=3
print(f'1*{j}={1*j}',end=',')
```

#  同上(加入迴圈)


```python
for i in range(1, 6):
    print(f'1*{j}={1*j}',end=',')
print()
for i in range(1, 6):
    print(f'2*{j}={2*j}',end=',')  
print()
for i in range(1, 6):
    print(f'3*{j}={2*j}',end=',')
print()
```

#  再加一層迴圈





```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```
