# 簡介

串列 (list) 是一個可以儲存多個元素 (整數，浮點數，或字串) 的容器。

串列中每個元素都是使用逗點隔開，左右有中括號 []。

```python
# scores 是一個串列，裏面每個元素都是整數。
scores = [78, 29, 56, 80, 99]
print(scores)
```

## 索引值

我們可以使用索引值 (index) 來存取串列中的元素。

注意：索引值由 0 開始。

```python
print(scores[0]) # 印出第 1 個元素
print(scores[1]) # 印出第 2 個元素
```

## 串列內容的個數

len(a) 取得 a 這個串列中元素的個數。

len 是 length (長度) 的意思

```python
print(len(scores))
print(scores[4]) # 取得最後一個元素
print(scores[5-1]) # 取得最後一個元素
print(scores[len(scores)-1])
print(scores[-1])
```

## 新增元素

使用 `append()` 新增元素到串列的尾端

```python
print(scores)
scores.append(78)
scores.append(93)
print(scores)
```

```python
print(scores)
scores[2] = 70 # 將 70 指定給索引值為 2 的元素
print(scores)
```

```python
weights = [12]*6
print(weights)
```

## 走訪串列的三種寫法



```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
# 對於 scores 中的每個元素，一次取一個，放在 e 中
for e in scores:
    print(e)
```

```python
for i, e in enumerate(scores):
    print(i,e)
```

## 計算費伯納西數列的第 91 個元素

一個費伯納西數列如下：

1,1,2,3,5,8,...

請問第 91 個元素為何？

提示：  
令 $x$ 是一個費伯納西數列, $x[i]$ 是第 $i$ 個數字。

我們會得到 $x[n] = x[n-1] + x[n-2]$

```python
# 初始化串列，包含 91 個 0
fib = [0]*91
print(fib)
fib[0] = 1
fib[1] = 1
print(fib)
i=2
fib[i] = fib[i-2] + fib[i-1]
print(fib)
i=3
fib[i] = fib[i-2] + fib[i-1]
print(fib)
i=4
fib[i] = fib[i-2] + fib[i-1]
print(fib)
```

```python
for i in range(2,91):
    fib[i] = fib[i-2] + fib[i-1]

print(fib[90])
```
